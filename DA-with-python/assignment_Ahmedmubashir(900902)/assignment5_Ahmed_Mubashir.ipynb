# QUESTION 1 SOLUTION

```python
# Create a list of 8 integers
numbers = [10, 15, 22, 7, 18, 25, 30, 11]

# Variable to store the total sum
total = 0

print("All numbers in the list:")

# Loop through each number in the list
for number in numbers:
    print(number)
    total = total + number

# Print the total sum
print("\nSum of all numbers:", total)

print("\nEven numbers:")

# Check and print only even numbers
for number in numbers:
    if number % 2 == 0:
        print(number)
```

# QUESTION 2 SOLUTION

```python
# Create a dictionary containing student information
student = {
    "name": "Ahmed",
    "age": 20,
    "course": "Python",
    "marks": [98, 92, 99]
}

print("Student Information:")

# Print all keys and their values
for key, value in student.items():
    print(key, ":", value)

# Variable to store the total marks
total = 0

# Add all marks using a loop
for mark in student["marks"]:
    total = total + mark

# Calculate the average marks
average = total / len(student["marks"])

print("\nAverage Marks:", average)

# Check whether the student passed or failed
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

# QUESTION 3 SOLUTON

```python
# Create a list containing employee dictionaries
employees = [
    {"name": "sameer", "department": "IT", "salary": 60000},
    {"name": "Ahmed", "department": "HR", "salary": 1000000},
    {"name": "usman", "department": "Finance", "salary": 70000}
]

print("Employee Details:")

# Variables to store the highest salary and total salary
highest_salary = 0
highest_employee = ""
total_salary = 0

# Loop through all employees
for employee in employees:
    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print()

    # Add each employee's salary to the total
    total_salary = total_salary + employee["salary"]

    # Check which employee has the highest salary
    if employee["salary"] > highest_salary:
        highest_salary = employee["salary"]
        highest_employee = employee["name"]

# Print the employee with the highest salary
print("Employee with highest salary:", highest_employee)
print("Highest salary:", highest_salary)

# Print the total salary expense
print("Total salary expense:", total_salary)
```

# QUESTION 4 SOLUTION 

```python
# Create an empty list to store the numbers
numbers = []

# Keep asking the user for numbers
while True:
    number = int(input("Enter a number (-1 to stop): "))

    # Stop the loop when the user enters -1
    if number == -1:
        break

    # Add the number to the list
    numbers.append(number)

# Print the complete list
print("\nComplete list:", numbers)

# Check if the list contains any numbers
if len(numbers) > 0:
    # Start with the first number as largest and smallest
    largest = numbers[0]
    smallest = numbers[0]

    # Compare all numbers in the list
    for number in numbers:
        if number > largest:
            largest = number

        if number < smallest:
            smallest = number

    # Print the largest and smallest numbers
    print("Largest number:", largest)
    print("Smallest number:", smallest)
else:
    # Display a message if no number was entered
    print("No numbers were entered.")
```

# QUESTION 5 SOLUTION


```python
# Ask the user to enter a sentence
sentence = input("Enter a sentence: ")

# Convert the sentence into lowercase words
words = sentence.lower().split()

# Create an empty dictionary to store word counts
frequency = {}

# Loop through each word
for word in words:
    # Check if the word already exists in the dictionary
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        # Add the word with a count of 1
        frequency[word] = 1

print("\nWord Frequency:")

# Print each word and its frequency
for word, count in frequency.items():
    print(word, ":", count)
```
