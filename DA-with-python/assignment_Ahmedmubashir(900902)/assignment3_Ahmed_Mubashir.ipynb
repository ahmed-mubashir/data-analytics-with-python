# TASK 2

```python
numbers = []

n = int(input("How many elements do you want to enter? "))

for i in range(n):
    numbers.append(input("Enter element: "))

new_list = []

for i in range(n - 1, -1, -1):
    new_list.append(numbers[i])

print("Reversed list:", new_list)

```

# TASK 3

```python
numbers = []

n = int(input("How many numbers do you want to enter? "))

for i in range(n):
    numbers.append(int(input("Enter number: ")))

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number is:", largest)

```

# TASK 4

```python
numbers = []

n = int(input("How many elements do you want to enter? "))

for i in range(n):
    numbers.append(input("Enter element: "))

last = numbers[-1]

for i in range(n - 1, 0, -1):
    numbers[i] = numbers[i - 1]

numbers[0] = last

print("Rotated list:", numbers)

```

```python
sentence = input("Enter a string: ")
word = input("Enter the word you want to delete: ")

words = sentence.split()
result = ""

for item in words:
    if item != word:
        result = result + item + " "

print("New string:", result.strip())

```

# TASK 7

```python
def capitalize_words(sentence):
    words = sentence.split()
    result = ""

    for word in words:
        result = result + word[0].upper() + word[1:] + " "

    return result.strip()

sentence = input("Enter a sentence: ")

print(capitalize_words(sentence))

```

```python
m = int(input("Enter rows: "))
n = int(input("Enter columns: "))

matrix = []

for i in range(m):
    row = []
    for j in range(n):
        row.append(int(input("Enter number: ")))
    matrix.append(row)

for i in range(m):
    total = 0
    for j in range(n):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)

```

# TASK 9

```python
n = int(input("Enter number of rows: "))
m = int(input("Enter number of columns: "))

matrix1 = []
matrix2 = []

print("Enter first matrix:")

for i in range(n):
    row = []
    for j in range(m):
        row.append(int(input("Enter number: ")))
    matrix1.append(row)

print("Enter second matrix:")

for i in range(n):
    row = []
    for j in range(m):
        row.append(int(input("Enter number: ")))
    matrix2.append(row)

result = []

for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("Sum of two matrices:")

for i in range(n):
    print(result[i])

```

# TASK 10


```python
r1 = int(input("Enter rows of first matrix: "))
c1 = int(input("Enter columns of first matrix: "))

r2 = int(input("Enter rows of second matrix: "))
c2 = int(input("Enter columns of second matrix: "))

if c1 != r2:
    print("Matrix multiplication is not possible")
else:
    matrix1 = []
    matrix2 = []

    print("Enter first matrix:")

    for i in range(r1):
        row = []
        for j in range(c1):
            row.append(int(input("Enter number: ")))
        matrix1.append(row)

    print("Enter second matrix:")

    for i in range(r2):
        row = []
        for j in range(c2):
            row.append(int(input("Enter number: ")))
        matrix2.append(row)

    result = []

    for i in range(r1):
        row = []
        for j in range(c2):
            total = 0
            for k in range(c1):
                total = total + matrix1[i][k] * matrix2[k][j]
            row.append(total)
        result.append(row)

    print("Multiplication of two matrices:")

    for i in range(r1):
        print(result[i])

```
